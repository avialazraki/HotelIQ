```python

import requests
import pandas as pd

print("Libraries imported successfully!")
```

```python

# Historical weather API for Lisbon, Portugal
url = "https://archive-api.open-meteo.com/v1/archive"

parameters = {
    "latitude": 38.72,
    "longitude": -9.14,
    "start_date": "2015-07-01",
    "end_date": "2017-08-31",
    "daily": "temperature_2m_max,temperature_2m_min,precipitation_sum",
    "timezone": "Europe/Lisbon"
}

# Request data from the API
response = requests.get(url, params=parameters, timeout=30)

# Check whether the request succeeded
print("Status code:", response.status_code)
response.raise_for_status()
```

```python

# Convert API response to JSON
weather_data = response.json()

# Display the main sections of the JSON response
print("JSON sections:")
print(weather_data.keys())

# Display the daily weather variables
print("\nDaily weather variables:")
print(weather_data["daily"].keys())
```

```python

# Convert daily weather JSON into a Pandas DataFrame
weather_df = pd.DataFrame(weather_data["daily"])

# Convert the date column to datetime format
weather_df["time"] = pd.to_datetime(weather_df["time"])

# Display the first 5 rows
weather_df.head()
```

```python

# Check the size of our dataset
print("Total rows:", len(weather_df))
print("Total columns:", len(weather_df.columns))

# Check for missing values
print("\nMissing values per column:")
print(weather_df.isnull().sum())

# Check for duplicate dates
print("\nDuplicate dates:", weather_df["time"].duplicated().sum())
```

```python

# Save historical weather data as CSV
weather_df.to_csv("HotelIQ_Lisbon_Weather.csv", index=False)

print("Weather CSV saved successfully!")
```
